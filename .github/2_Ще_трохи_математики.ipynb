```python
from math import sqrt,pi
S = float(input('Введіть площу кола: '))
r = sqrt(S/pi)
print ('Радіус дорівнює %6.3f см' %r)
```

```python
from math import pi
R = float(input('Введіть радіус кола: '))
S = 4 * pi * R**2
V = (4/3) * pi * R**3
print ('Площа дорівнює %6.3f см²' %S)
print ('Об\'єм дорівнює %6.3f см³' %V)
```

```python
choice = input("Конвертація (1 - F, 2 - C у F): ")
if choice == '1':
    F = float(input("Введіть температуру у Френгейтах: "))
    C = (F - 32) * 5/9
    print(f"{C} C = {F:.2f} F")
elif choice == '2':
    C = float(input("Введіть температуру у Цельсіях: "))
    F = C * 9/5 + 32
    print(f"{C} C = {F:.2f} F")
else:
    print("Невірний вибір.")
```
